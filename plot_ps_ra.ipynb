```python
import yaml
import matplotlib.pyplot as plt

from src import plot_ra_ps_summary

# YAMLの読み込み
yaml_path = "data/ps_ra_data.yaml"
with open(yaml_path, "r", encoding="utf-8") as f:
    payload = yaml.safe_load(f)

# データ抽出
series_list = []
labels = []
colors = []

root = payload.get("data", {})
for group_name, group in root.items():
    # group: measured_in_2025 / measured_in_2026 など
    for key, item in group.items():
        data = item.get("data", [])
        label = item.get("label", key)
        color = item.get("color", "tab:blue")
        if data:
            series_list.append(data)
            labels.append(label)
            colors.append(color)

# プロット
fig, ax = plot_ra_ps_summary(
    series_list=series_list,
    labels=labels,
    colors=colors,
    title="RA vs Ps",
    ylim=(0, 30)
)
plt.show()

```
